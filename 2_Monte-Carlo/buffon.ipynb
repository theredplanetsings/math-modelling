## Buffon's Needle 

Buffon's Needle Experiments

Description: This program simulates Buffon's Needle experiment in the clase where L = D = 1, L = 1/2 & D = 2, and L = 1 & D = 20

```python
import numpy as np
import math 

L = 1  # length of the needle
D = 1  # distance between the lines
n = 5000

print('=============================')
print('Buffon''s Needle Experiment')
print('=============================\n')

print('Total drops   Estimate for Pi')
print('=============================')

crossings = 0
for i in range(n):
  randomNumber = np.random.rand()
  verticalPosition = D*randomNumber
  anotherRandomNumber = np.random.rand()
  theta = math.pi*anotherRandomNumber
  if (verticalPosition <= L*np.sin(theta)):
    crossings += 1
  if (np.mod(i,np.floor(n/10))==0) and (crossings > 0): 
      print('%8d   %10.5f' % (i, 2*i*L/(D*crossings)))

if crossings > 0: 
    print('\nFinal Estimate after %d drops: %7.5f' % (n, 2*n*L/(D*crossings)))
else: 
    print('\n>>>> Ouch, we got no crossings which means no estimate due to a divide by zero!\n')
print('Length of Needle       = %f' % L)
print('Distance between Lines = %f\n' % D)
```

### L = 3, D = 4, Fox's experiment

```python
import numpy as np
import math 

L = 3  # length of the needle
D = 4  # distance between the lines
n = 530  # Fox's number of tosses
num_simulations = 100000  # Number of experiments to run
fox_estimate = 3.1423  # Fox's reported estimate

# Counter for experiments that achieve Fox's result or better
successful_experiments = 0

print('=============================')
print('Fox Experiment Analysis')
print('=============================\n')
print(f'Running {num_simulations} simulations of {n} tosses each...\n')

# Run many simulations
for sim in range(num_simulations):
    crossings = 0
    for i in range(n):
        randomNumber = np.random.rand()
        verticalPosition = D * randomNumber
        anotherRandomNumber = np.random.rand()
        theta = math.pi * anotherRandomNumber
        if (verticalPosition <= L * np.sin(theta)):
            crossings += 1
    
    # Calculate estimate for this simulation
    if crossings > 0:
        pi_estimate = 2 * n * L / (D * crossings)
        # Check if estimate is as good or better than Fox's
        # "Better" means closer to actual π (3.14159...)
        if abs(pi_estimate - math.pi) <= abs(fox_estimate - math.pi):
            successful_experiments += 1

# Calculate probability
probability = successful_experiments / num_simulations

print('Results:')
print('=============================')
print(f'Fox\'s estimate: {fox_estimate}')
print(f'Actual π: {math.pi:.5f}')
print(f'Error in Fox\'s estimate: {abs(fox_estimate - math.pi):.5f}')
print(f'\nSimulations achieving Fox\'s accuracy or better: {successful_experiments}')
print(f'Probability: {probability:.4f} or {probability*100:.2f}%')
print(f'\nConclusion: Fox\'s result was {"quite lucky" if probability < 0.10 else "reasonably likely" if probability < 0.30 else "fairly common"}.')
```
