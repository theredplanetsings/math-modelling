## Fish tank modeling - first pass

python program 

First pass at a python program for Monte Carlo simulation of the fish tank modeling problem.

```python
import numpy as np

a = 1/3   # probability of a customer each day
days_for_delivery = 2   # days from order to delivery of new tanks

stock = 1   # number of tanks in stock
deliv = -1  # number of days until delivery of tank on order
            # -1 means none on order

total_cust = 0
total_sold = 0
total_lost = 0

print('week day stock cust sold lost')

for week in range(3):
  for weekday in range(7):
    sold = 0
    lost = 0
    if deliv == 0:
      stock += 1   # a new tank is delivered
    if deliv >= 0:
      deliv -= 1   # days till next delivery

    random_num = np.random.rand()  # generate random number
    if random_num < a:             # use this number to tell if a customer arrived
       customers = 1
    else:
       customers = 0

    if customers==1:
       if stock>0:               # we have a tank to sell the customer
          sold += 1
          stock -= 1
          if deliv < 0:
             deliv = days_for_delivery  # we sold a tank and now order
                                        # another one.
       else:
          lost = lost+1      # we didn't have a tank and lost a customer

    # keep track of total statistics:
    total_cust += customers
    total_sold += sold
    total_lost += lost

    # display results for this day 
    print('%4d %3d %5d %4d %4d %4d' % (week, weekday, stock, \
                                       customers, sold, lost))   

# display final totals    
print('\ncustomers  sold  lost')
print('%8d    %2d   %2d' % (total_cust,total_sold,total_lost))
```
