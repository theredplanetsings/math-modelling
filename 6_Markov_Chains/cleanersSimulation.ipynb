## Monte Carlo Simulation for Cleaners Problem 
Introducing Markov Chains
    
Description: This code simulates the following: A small town has 
only two dry cleaners, Starch Cleaners and Spotless Cleaners. 
Starch Cleaner's manager wants to increase business with an 
advertising campaign.  A market research firm finds that after 
the marking campaign there is a 0.8 probability that a customer 
of Starch's will bring the next batch of dirty clothes to Starch 
and a 0.35 chance that a Spotless customer will switch to Starch
for the next batch of clothes.  

```python
import numpy as np

num_experiments = 10000   # num_experiments = number of simulations
time_periods    = 100     # number of time_periods to walk through
initial_state   = 1       # state to start the simulations

# The following matrix will keep track of which state (i.e. which cleaners)
# someone is using at each of the time_periods.  
time_state_matrix = np.zeros((time_periods, 2))

for i in range(num_experiments):
    # We begin each simulation with a customer being in State 1
    # This can be changed and will vary your results. 
    current_state = initial_state
    time_state_matrix[0,current_state] += 1
    for time_step in range(1,time_periods):
        rand_num = np.random.rand()
        if current_state == 1:   # In State 1 (Spotless)     
           if (rand_num <= 0.8): # Then customer stays with Splotless
              time_state_matrix[time_step,0] += 1
              current_state = 1
           else:   # customer switched
              time_state_matrix[time_step,1] += 1
              current_state = 2
        else:  # we are in State 2 (Starch)
           if (rand_num <= 0.65):
              time_state_matrix[time_step,1] += 1
              current_state = 2
           else:   # customer switched
              time_state_matrix[time_step,0] += 1
              current_state = 1

print('=========================================')
print('Spotless and Starch Cleaners Simulation')
print('=========================================')
print('Time Period   State 1     State 2')
print('=========================================')

for i in range(6):
    print('  %3d         %5d        %4d' % (i,time_state_matrix[i,0],time_state_matrix[i,1]))
    
print('\n')

for i in range(50,57): 
    print('  %3d         %5d        %4d' % (i,time_state_matrix[i,0],time_state_matrix[i,1]))

print('\n')

for i in range(94,100): 
    print('  %3d         %5d        %4d' % (i,time_state_matrix[i,0],time_state_matrix[i,1]))
```
